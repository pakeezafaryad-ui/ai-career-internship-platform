# AI Career and Internship Agent Platform — Step 1

This notebook prepares the starter project in Google Colab, installs its initial dependency, and runs a basic source check.

**Safety:** Do not upload real CVs or put API keys directly into notebook cells. This starter does not yet process private data or call an AI API.

```python
# Install the initial application dependency
%pip -q install 'streamlit>=1.40,<2.0'
print('Streamlit installation complete.')
```

## 1. Create the starter files

Run this cell to create the first `app.py` and `requirements.txt` directly in the Colab runtime.

```python
from pathlib import Path

PROJECT_DIR = Path('/content/ai-career-internship-platform')
PROJECT_DIR.mkdir(parents=True, exist_ok=True)

app_code = '''import streamlit as st
st.set_page_config(page_title="AI Career & Internship Agent", page_icon="🎓", layout="wide")
st.title("🎓 AI Career and Internship Agent Platform")
st.caption("A secure, multi-agent career guidance and internship matching project")
st.info("Starter demo only. Accounts, private CV processing, AI agents, and live vacancies are not implemented yet.")
col1, col2, col3 = st.columns(3)
with col1:
    st.markdown("### 📄 CV Analyzer")
    st.write("Planned: extract CV details and identify skills.")
with col2:
    st.markdown("### 🧭 Career Advisor")
    st.write("Planned: recommend career paths with explanations.")
with col3:
    st.markdown("### 💼 Job Matcher")
    st.write("Planned: compare skills with job requirements.")
interest = st.selectbox("Which area interests you most?", ["AI / Machine Learning", "Software Development", "Cybersecurity", "Data Analytics", "UI/UX Design"])
if st.button("Show starter guidance"):
    tips = {"AI / Machine Learning": "Start with Python, data handling, basic statistics, and small ML projects.", "Software Development": "Practice programming fundamentals, Git, testing, data structures, and one portfolio project.", "Cybersecurity": "Learn networking, OS basics, security principles, and legal lab-based practice.", "Data Analytics": "Start with spreadsheets, SQL, Python or a BI tool, and data visualization.", "UI/UX Design": "Practice user research, wireframes, accessibility, prototyping, and portfolio case studies."}
    st.success(tips[interest])
st.caption("Demo suggestions are general guidance, not guarantees of employment.")
'''

(PROJECT_DIR / 'app.py').write_text(app_code, encoding='utf-8')
(PROJECT_DIR / 'requirements.txt').write_text("streamlit>=1.40,<2.0\\n", encoding='utf-8')
print('Created:', PROJECT_DIR)
print('Files:', [p.name for p in PROJECT_DIR.iterdir()])
```

## 2. Check the Python source

This compiles `app.py` without launching the web server.

```python
source = (PROJECT_DIR / 'app.py').read_text(encoding='utf-8')
compile(source, str(PROJECT_DIR / 'app.py'), 'exec')
print('PASS: app.py has valid Python syntax.')
```

## 3. Download starter files

This creates a ZIP of the notebook-created starter files so you can download them from Colab.

```python
import zipfile
from google.colab import files

zip_path = '/content/ai-career-internship-starter-colab.zip'
with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zf:
    zf.write(PROJECT_DIR / 'app.py', 'ai-career-internship-platform/app.py')
    zf.write(PROJECT_DIR / 'requirements.txt', 'ai-career-internship-platform/requirements.txt')
files.download(zip_path)
```

## 4. Optional: run Streamlit inside Colab

This starts Streamlit and exposes it using a temporary tunnel. A public tunnel may expose the demo page to others, so do not use this method for private user data or secrets. For a shareable project demo, we will later deploy through a properly configured hosting workflow.

```python
# Optional only: run if you want to preview the starter UI in Colab.
# The tunnel URL is public while active. Do not put sensitive data in this demo.
%pip -q install pyngrok
from pyngrok import ngrok
import subprocess, time, os

# Add your ngrok auth token through Colab Secrets if you choose to use this.
# Do not paste a token directly into a notebook cell.
from google.colab import userdata
token = userdata.get('NGROK_AUTH_TOKEN')
if not token:
    print('Optional preview skipped. Add NGROK_AUTH_TOKEN in Colab Secrets to enable the tunnel.')
else:
    ngrok.set_auth_token(token)
    proc = subprocess.Popen(['streamlit', 'run', str(PROJECT_DIR / 'app.py'), '--server.port', '8501', '--server.address', '0.0.0.0'], stdout=subprocess.DEVNULL, stderr=subprocess.STDOUT)
    time.sleep(3)
    public_url = ngrok.connect(8501, 'http').public_url
    print('Temporary demo URL:', public_url)
    print('Stop the runtime when finished. Do not enter private information into this public demo.')
```
